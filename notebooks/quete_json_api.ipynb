```python
import requests

URL_ADRESSE = "https://api-adresse.data.gouv.fr/search/"
ENTETES = {"User-Agent": "bootcamp-data/1.0 (github.com/noah-mathieu-data)"}

r = requests.get(
    URL_ADRESSE,
    params={"q": "20 avenue de Ségur 75007 Paris", "limit": 1},
    headers=ENTETES,
    timeout=10,
)

print("status :", r.status_code)

premier = r.json()["features"][0]["properties"]
print("label :", premier["label"])
print("score :", premier["score"])
```

```python
adresses = [
    "20 avenue de Ségur 75007 Paris",
    "Grand Place, Lille",
    "1 rue de la Paix, Paris",
    "place Bellecour, Lyon",
    "zzzzz qwerty 99999",
]

def interroger(url, params, headers=None):
    try:
        r = requests.get(url, params=params, headers=headers, timeout=10)
        r.raise_for_status()
        return r.json()
    except requests.exceptions.Timeout:
        print("Trop lent, on abandonne cette requête")
    except requests.exceptions.HTTPError as e:
        print("Le serveur a répondu une erreur :", e)
    except requests.exceptions.RequestException as e:
        print("Problème réseau :", e)
    return None


resultats = []
for adresse in adresses:
    reponse = interroger(URL_ADRESSE, {"q": adresse, "limit": 1}, ENTETES)
    if reponse and reponse["features"]:
        p = reponse["features"][0]["properties"]
        resultats.append({
            "adresse": adresse,
            "label": p["label"],
            "score": round(p["score"], 2),
            "postcode": p.get("postcode"),
            "city": p.get("city"),
        })
    else:
        resultats.append({"adresse": adresse, "label": None, "score": None,
                          "postcode": None, "city": None})

df_adresses = pd.DataFrame(resultats)
df_adresses
```

```python
print("adresses trouvées :", df_adresses["label"].notna().sum(), "sur", len(df_adresses))
print("score le plus bas :", df_adresses["score"].min())
print(df_adresses.loc[df_adresses["score"].idxmin(), ["adresse", "label"]])
```
