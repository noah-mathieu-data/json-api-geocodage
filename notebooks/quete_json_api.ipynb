```python
import json

with open("../data/villes.json", encoding="utf-8") as f:
    villes = json.load(f)

print("nombre de villes :", len(villes["villes"]))
print("2e ville :", villes["villes"][1]["nom"])
print("restaurants à Syracuse :", len(villes["villes"][2]["restaurants"]))
print("1er restaurant de Palerme :", villes["villes"][1]["restaurants"][0]["nom"])
```

```python
for v in villes["villes"]:
    print(f"{v['nom']:10} {len(v['restaurants'])} restaurants")
```

```python
import pandas as pd

df_villes = pd.json_normalize(
    villes,
    record_path=["villes", "restaurants"],
    meta=[["villes", "nom"], ["villes", "region"], ["villes", "habitants"]],
)

df_villes = df_villes.rename(columns={
    "villes.nom": "ville",
    "villes.region": "region",
    "villes.habitants": "habitants",
})

print(df_villes.shape)
df_villes
```

```python
df_villes["note_moyenne"] = df_villes["notes"].apply(lambda x: round(sum(x) / len(x), 2))

moyenne_ville = df_villes.groupby("ville")["note_moyenne"].mean().round(2).sort_values(ascending=False)
print(moyenne_ville)
print()

print(df_villes["type"].value_counts())

df_villes
```
